##### ----------------------------------------------------------

##### 1. Verifica indirizzo MAC

```python
import re

def is_mac_address(s: str) -> bool:
    pattern = r'^([0-9A-Fa-f]{2}:){5}[0-9A-Fa-f]{2}$'
    return re.fullmatch(pattern, s) is not None
```

##### ----------------------------------------------------------

##### 2. Estrai parole SOLO MAIUSCOLE (minimo 2 lettere)

```python
import re

def extract_uppercase_words(text: str):
    pattern = r'\b[A-Z]{2,}\b'
    return re.findall(pattern, text)
```
##### ----------------------------------------------------------

##### 3. Verifica orario HH:MM (00–23, 00–59)

```python
import re

def is_valid_time(s: str) -> bool:
    pattern = r'^(?:[01]\d|2[0-3]):[0-5]\d$'
    return re.fullmatch(pattern, s) is not None
```

##### ----------------------------------------------------------

##### 4. Estrai gruppi di cifre da stringa mista

```python
import re

def extract_numbers(text: str):
    pattern = r'\d+'
    return [int(n) for n in re.findall(pattern, text)]
```

##### ----------------------------------------------------------

##### 5. Verifica prezzo nel formato "€ ##,##"

##### - Almeno una cifra intera
##### - Decimali opzionali
##### - Virgola come separatore decimale

```python
import re

def is_price(s: str) -> bool:
    pattern = r'^€\s?\d+(?:,\d{2})?$'
    return re.fullmatch(pattern, s) is not None
```

##### ----------------------------------------------------------

##### 6. Estrai parole con iniziale maiuscola da testo con più paragrafi

```python
import re

def extract_capitalized_words(text: str):
    pattern = r'\b[A-Z][a-zA-Z]*\b'
    return re.findall(pattern, text)
```

```python
if __name__ == "__main__":
    print(is_mac_address("AA:00:CC:DD:99:FF"))
    print(extract_uppercase_words("CIAO a tutti dagli USA"))
    print(is_valid_time("23:59"))
    print(extract_numbers("ID45A23X"))
    print(is_price("€ 12,50"))
    print(extract_capitalized_words("Ciao. Questo è un Test."))
```
